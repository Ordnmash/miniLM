```python
import json
data = None
with open('external_dataset.json', 'r') as f:
  data = json.load(f)
```

```python
dataset = []
```

```python
# unpack the entire json monster and rewrite formatting with special tokens
for d in data: # d == obj
  for j in d['turns']:
    dataset.append('<start_of_turn>user' if j['speaker'] ==
                   'user' else '<start_of_turn>model')
    dataset.append(j['utterance']+'<end_of_turn>')
```

```python
gg = open('converted_chat_formatted.txt', 'r').read().splitlines() # copy for the same dataset
```

```python
gg[:30]
```

```python
with open('converted_chat_formatted.txt', 'w') as fille:
  for line in gg + dataset:
    fille.write(line)
    fille.write("\n")
```
